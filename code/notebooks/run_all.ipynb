# MBC R3 offline reproduction

This notebook calls the single offline entry point. It does not download data.

```python
from pathlib import Path
import os, subprocess, sys
ROOT = Path.cwd()
while not (ROOT / 'code' / 'run_all.py').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
assert (ROOT / 'code' / 'run_all.py').exists(), 'Open this notebook inside the package or a subdirectory.'
env = os.environ.copy()
env['MBC_OFFLINE'] = '1'
env['MPLBACKEND'] = 'Agg'
subprocess.run([sys.executable, str(ROOT / 'code' / 'run_all.py')], cwd=ROOT, env=env, check=True)
```

```python
import json
summary = ROOT / 'results' / 'generated_json' / 'run_all_manifest.json'
print(json.loads(summary.read_text(encoding='utf-8')))
```
